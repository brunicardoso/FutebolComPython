## 🎯 Missão 1: Quantos gols teve o jogo?

O Brasil ganhou da Argentina por **3 x 1**.

Cada número fica guardado numa **variável**: uma caixinha com nome. `gols_brasil` guarda 3, `gols_argentina` guarda 1.

🤔 **Palpite:** quantos gols teve o jogo, somando os dois times? Pense antes de rodar!

✏️ Troque o `___` pelo sinal de **mais** (`+`) e aperte play.

```python
gols_brasil = 3
gols_argentina = 1
total = gols_brasil ___ gols_argentina
print(total)
checar(total, "m1")
```

🤔 **Pense:** se a Argentina tivesse feito mais 2 gols, quantos gols teria o jogo?

✏️ **Tente:** mude `gols_argentina = 1` para `gols_argentina = 3` e rode de novo. (Se aparecer um erro vermelho, é porque ainda tem `___` sobrando.)

## 🎯 Missão 2: Saldo de gols

No campeonato, o **saldo de gols** é: gols que o time **fez** menos gols que o time **sofreu**.

O Tubarões FC fez **10 gols** e sofreu **14 gols**.

🤔 **Palpite:** o saldo vai ser positivo (maior que zero) ou negativo (menor que zero)?

✏️ Troque o `___` pelo sinal de **menos** (`-`).

```python
gols_feitos = 10
gols_sofridos = 14
saldo = gols_feitos ___ gols_sofridos
print(saldo)
checar(saldo, "m2")
```

🤔 **Pense:** o resultado foi **−4**. Um número com o sinal de menos na frente é um **número negativo**: ele é menor que zero. É como um termômetro no frio: 4 graus abaixo de zero!

Saldo negativo quer dizer que o time sofreu mais gols do que fez.

✏️ **Tente:** quantos gols o Tubarões FC precisaria fazer para o saldo ficar **zero**? Mude `gols_feitos` e confira.

## 🎯 Missão 3: Pontos no campeonato

No campeonato, cada **vitória vale 3 pontos**. O Foguete EC ganhou **8 jogos**.

Para multiplicar, o Python usa a **estrelinha** `*` (e não o ×).

✏️ Troque o `___` pela estrelinha `*`.

```python
vitorias = 8
pontos = vitorias ___ 3
print(pontos)
checar(pontos, "m3")
```

🤔 **Pense:** se o time também empatou 2 jogos (cada empate vale 1 ponto), quantos pontos ele tem no total?

✏️ **Tente:** escreva `print(8 * 3 + 2 * 1)` numa célula nova e confira.

## 🎯 Missão 4: Gols por jogo

Um atacante fez **30 gols** em **15 jogos**. Quantos gols ele fez **por jogo**, em média?

Para dividir, o Python usa a **barra** `/`.

✏️ Troque o `___` pela barra `/`.

```python
gols = 30
jogos = 15
gols_por_jogo = gols ___ jogos
print(gols_por_jogo)
checar(gols_por_jogo, "m4")
```

🤔 **Pense:** o Python mostrou `2.0` e não `2`. Ele usa **ponto** no lugar da vírgula, e a divisão sempre vem com um número depois do ponto. `2.0` é o mesmo que 2.

✏️ **Tente:** e se fossem 30 gols em 20 jogos? O resultado `1.5` quer dizer um gol e meio por jogo.
