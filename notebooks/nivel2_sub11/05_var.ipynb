## 🔥 Aquecimento

O **VAR** é o árbitro de vídeo. Ele olha o lance e responde perguntas de **sim ou não**.

No Python, uma pergunta de sim ou não vira **`True`** (verdadeiro, sim) ou **`False`** (falso, não).

- `>` pergunta: é **maior que**?
- `7 > 5` pergunta: 7 é maior que 5?

Aperte play.

````python
print(7 > 5)
checar(7 > 5, "aquecimento")
````

- `<` pergunta: é **menor que**?
- `==` pergunta: é **igual**? (Lembre: um `=` guarda, dois `==` perguntam.)

🤔 **Palpite:** o time da casa ganhou? Vai dar `True` ou `False`?

✏️ Troque o `___` pelo sinal que pergunta **"é maior que?"**.

````python
gols_casa = 2
gols_fora = 1
venceu = gols_casa ___ gols_fora
print(venceu)
checar(venceu, "m1")
````

🤔 **Pense:** `venceu` guardou `True`. Uma variável pode guardar um número, um texto **ou** uma resposta de sim ou não.

## 🎯 Missão 2: Dentro ou fora da área?

Uma falta **dentro da área** vira **pênalti**. Fora da área, é só **falta**.

O `if` (*se*) faz o Python escolher um caminho:

```
if pergunta:
    faz isto quando é True
else:
    faz isto quando é False
```

- `else` quer dizer **senão**.
- As linhas de dentro começam com **4 espaços**. É assim que o Python sabe o que fica dentro do `if`.

Esta falta foi **fora** da área (`dentro_da_area = False`). ✏️ Troque o `___` pela decisão certa: `falta`.

````python
dentro_da_area = False
if dentro_da_area:
    decisao = "pênalti"
else:
    decisao = "___"
print("O VAR diz:", decisao)
checar(decisao, "m2")
````

🤔 **Pense:** como `dentro_da_area` é `False`, o Python pulou o caminho do `if` e foi para o `else`.

✏️ **Tente:** troque para `dentro_da_area = True` e rode de novo.

## 🎯 Missão 3: Mão na bola (regra com E)

Na nossa regra, mão na bola só é pênalti se as **duas** coisas forem verdade:

- a mão foi **dentro da área**, **E**
- o **braço estava aberto**.

`and` quer dizer **E**: só dá `True` se os dois lados forem `True`.

Neste lance, foi dentro da área, mas o braço estava **colado no corpo**.

🤔 **Palpite:** é pênalti?

✏️ Troque o `___` por `and`.

````python
dentro_da_area = True
braco_aberto = False
penalti = dentro_da_area ___ braco_aberto
print("Foi pênalti?", penalti)
checar(penalti, "m3")
````

🤔 **Pense:** com `and`, basta **um** lado ser `False` para a resposta ser `False`.

| dentro da área | braço aberto | pênalti? (`and`) |
|---|---|---|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

## 🎯 Missão 4: Cartão vermelho (regra com OU)

O jogador leva **vermelho** se a falta foi **violenta**, **OU** se foi o **segundo amarelo** dele.

`or` quer dizer **OU**: dá `True` se **pelo menos um** lado for `True`.

Neste lance, a falta não foi violenta, mas era o **segundo amarelo**.

✏️ Troque o `___` por `or`.

````python
violenta = False
segundo_amarelo = True
vermelho = violenta ___ segundo_amarelo
print("Cartão vermelho?", vermelho)
checar(vermelho, "m4")
````

🤔 **Pense:** com `or`, basta **um** lado ser `True`. Só dá `False` quando os dois são `False`.

✏️ **Tente:** troque `or` por `and` e veja a resposta mudar. Qual regra é mais justa para esse lance?

## 🧢 Desafio do Técnico

**Desafio 1:** um jogo pode ter **três** resultados: vitória, empate ou derrota. Para três caminhos, usamos `elif` (é um *senão, se*).

O jogo foi **1 x 1**. ✏️ Troque o `___` pelo sinal que pergunta **"é igual?"**.

````python
gols_casa = 1
gols_fora = 1
if gols_casa > gols_fora:
    resultado = "vitória"
elif gols_casa ___ gols_fora:
    resultado = "empate"
else:
    resultado = "derrota"
print(resultado)
checar(resultado, "desafio1")
````
