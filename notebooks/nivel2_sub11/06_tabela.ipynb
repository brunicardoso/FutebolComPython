## 🎯 Missão 1: Somando a campanha

Os **Tigres FC** jogaram 5 jogos. A lista `resultados` guarda o que aconteceu, em ordem.

- `for r in resultados:` quer dizer: **para cada** resultado `r` da lista, faça o que está embaixo.
- `pontos` começa em 0 e vai **acumulando**. Por isso se chama **acumulador**.

🤔 **Palpite:** quantos pontos os Tigres fizeram? Conte de cabeça: V, E, V, D, V.

✏️ O empate vale quanto? Troque o `___`.

```python
resultados = ["V", "E", "V", "D", "V"]
pontos = 0
for r in resultados:
    if r == "V":
        pontos = pontos + 3
    elif r == "E":
        pontos = pontos + ___
print("Pontos dos Tigres:", pontos)
checar(pontos, "m1")
```

🤔 **Pense:** a derrota não tem linha nenhuma. Por quê? Porque somar 0 não muda nada!

✏️ **Tente:** troque a lista por `["D", "D", "V"]` e rode de novo.

## 🎯 Missão 2: Contando vitórias

Um atalho: `pontos += 3` é o mesmo que `pontos = pontos + 3`. (`+=` quer dizer **some e guarde**.)

Agora vamos contar **quantas vitórias** os Tigres tiveram. Cada vitória soma **1** no contador.

✏️ Troque o `___`.

```python
resultados = ["V", "E", "V", "D", "V"]
vitorias = 0
for r in resultados:
    if r == "V":
        vitorias += ___
print("Vitórias:", vitorias)
checar(vitorias, "m2")
```

🤔 **Pense:** contar é somar 1 cada vez que acontece. Somar 3 cada vez dá os pontos das vitórias: 3 vitórias × 3 = 9.

## 🎯 Missão 3: Saldo de gols

**Saldo de gols** = gols feitos − gols sofridos.

As duas listas guardam os gols dos 5 jogos. O primeiro `for` soma os gols feitos. O segundo precisa **tirar** os gols sofridos.

- `-=` quer dizer **subtraia e guarde**.

🤔 **Palpite:** qual vai ser o saldo?

✏️ Troque o `___` por `-=`.

```python
gols_feitos = [2, 1, 3, 0, 2]
gols_sofridos = [0, 1, 1, 2, 1]
saldo = 0
for g in gols_feitos:
    saldo += g
for g in gols_sofridos:
    saldo ___ g
print("Saldo de gols:", saldo)
checar(saldo, "m3")
```

🤔 **Pense:** os Tigres fizeram 8 gols e sofreram 5. O saldo pode ser negativo também, igual no notebook 01!
