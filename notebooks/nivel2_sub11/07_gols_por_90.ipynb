## 🔥 Aquecimento

Um jogo de futebol tem **90 minutos**. Um atacante fez **2 gols** em **180 minutos** (dois jogos inteiros).

Quantos gols ele faz **a cada 90 minutos**? Divida os gols pelos minutos e multiplique por 90. Aperte play.

```python
jogadores = carregar_jogadores()
print(2 / 180 * 90)
checar(2 / 180 * 90, "aquecimento")
```

## 🎯 Missão 1: Uma coluna nova

Vamos fazer essa conta para **todos** os jogadores de uma vez, criando uma coluna nova chamada `gols_por_90`.

- `jogadores["gols"] / jogadores["minutos"]` divide os gols pelos minutos, linha por linha.
- Escrever `jogadores["gols_por_90"] = ...` **cria** a coluna nova na tabela.

✏️ Troque o `___` pelo número de minutos de um jogo.

```python
jogadores["gols_por_90"] = jogadores["gols"] / jogadores["minutos"] * ___
print(jogadores[["nome", "gols", "minutos", "gols_por_90"]].head())
checar(jogadores["gols_por_90"].iloc[0], "m1")
```

🤔 **Pense:** `iloc[0]` pega a **primeira linha** (a posição 0, lembra das listas?). O Harry Kane faz mais ou menos 1,36 gol a cada 90 minutos. Isso é mais de um gol por jogo!
