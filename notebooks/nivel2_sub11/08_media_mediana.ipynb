## 🎯 Missão 1: O mais caro e o mais barato

A coluna `valor_milhoes` guarda quanto cada jogador vale, em **milhões de euros**.

- `.max()` dá o **maior** valor. (*max* vem de *máximo*.)
- `.min()` dá o **menor**. (*min* vem de *mínimo*.)

🤔 **Palpite:** quanto vale o jogador mais caro da tabela?

✏️ Troque o `___` por `max`.

```python
mais_caro = jogadores["valor_milhoes"].___()
print("O mais caro vale", mais_caro, "milhões")
print("O mais barato vale", jogadores["valor_milhoes"].min(), "milhões")
checar(mais_caro, "m1")
```

## 🎯 Missão 2: A média de um time pequeno

Um time da base tem 5 jogadores. Os valores deles (em milhões) estão na lista `valores`.

- `import statistics` traz a caixa de ferramentas de **estatística** do Python.
- `statistics.mean(lista)` calcula a **média**. (*mean* quer dizer *média*.)

🤔 **Palpite:** qual é a média?

✏️ Troque o `___` por `mean`.

```python
import statistics
valores = [2, 3, 4, 5, 6]
media = statistics.___(valores)
print("Média:", media)
checar(media, "m2")
```

🤔 **Pense:** 2 + 3 + 4 + 5 + 6 = 20, e 20 ÷ 5 = 4. A média é 4 milhões.

## 🎯 Missão 3: Chegou um supercraque!

O time contratou um craque que vale **180 milhões**! Vamos ver o que acontece com a média e com a **mediana**.

A **mediana** é o valor **do meio** quando a lista está em ordem. Metade fica abaixo, metade fica acima. Com 6 valores, ela é a média dos dois do meio.

- `statistics.median(lista)` calcula a mediana.

✏️ Troque o `___` por `median`.

```python
valores = [2, 3, 4, 5, 6, 180]
print("Nova média:", statistics.mean(valores))
mediana = statistics.___(valores)
print("Mediana:", mediana)
checar(mediana, "m3")
```
